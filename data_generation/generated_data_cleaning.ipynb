```python
import json
import pandas as pd
from collections import Counter

# Path to the jsonl file
file_path = "../../Data/Finetuning/Augmented/Small/Other/dolly_train_3_Qwen2.5-7B-Instruct.jsonl"

# Read the jsonl file
data = []
with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        data.append(json.loads(line))

# Extract instructions
instructions = [entry["instruction"] for entry in data]

# Check for duplicates
instruction_counts = Counter(instructions)
duplicates = {instruction: count for instruction, count in instruction_counts.items() if count > 1}

# Display results
print(f"Total entries: {len(data)}")
print(f"Unique instructions: {len(instruction_counts)}")
print(f"Number of instructions with duplicates: {len(duplicates)}")

if duplicates:
    print("\nDuplicate instructions:")
    for instruction, count in sorted(duplicates.items(), key=lambda x: x[1], reverse=True):
        print(f"Count: {count}, Instruction: {instruction[:100]}..." if len(instruction) > 100 else f"Count: {count}, Instruction: {instruction}")
    
    # Create a DataFrame for easier analysis
    df = pd.DataFrame(data)
    
    # Example: Get the first duplicate instruction
    if duplicates:
        first_duplicate = list(duplicates.keys())[0]
        print(f"\nExample of duplicate entries for: {first_duplicate[:100]}..." if len(first_duplicate) > 100 else f"\nExample of duplicate entries for: {first_duplicate}")
        duplicate_entries = df[df['instruction'] == first_duplicate]
        for i, entry in duplicate_entries.iterrows():
            print(f"\nEntry {i+1}:")
            print(f"Response: {entry['response_Qwen2.5-7B-Instruct'][:200]}...")
```

```python
# Remove duplicate entries
print("\nRemoving duplicate entries...")

# Create a list to store unique entries
unique_data = []
seen_instructions = set()

# Keep only the first occurrence of each instruction
for entry in data:
    instruction = entry["instruction"]
    if instruction not in seen_instructions:
        unique_data.append(entry)
        seen_instructions.add(instruction)

# Print results
print(f"Original data count: {len(data)}")
print(f"After removing duplicates: {len(unique_data)}")
print(f"Removed {len(data) - len(unique_data)} duplicate entries")

# Save the deduplicated data to a new file
output_file_path = file_path.replace('.jsonl', '_deduplicated.jsonl')

with open(output_file_path, 'w', encoding='utf-8') as f:
    for entry in unique_data:
        f.write(json.dumps(entry) + '\n')

print(f"Deduplicated data saved to: {output_file_path}")
```
