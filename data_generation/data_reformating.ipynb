```python
import json
import os
from pathlib import Path
import shutil
```

```python
def process_jsonl_file(input_file, output_file):
    with open(input_file, 'r', encoding='utf-8') as f_in, \
         open(output_file, 'w', encoding='utf-8') as f_out:
        for line in f_in:
            data = json.loads(line.strip())
            
            # Find the model response key
            model_key = None
            for key in data.keys():
                if key.startswith('response_') and key != 'response_human':
                    model_key = key
                    break
            
            if model_key:
                # Extract model name from the key
                model_name = model_key.replace('response_', '')
                
                # Create new data structure
                new_data = {
                    'instruction': data['instruction'],
                    'response_human': data['response_human'],
                    'response_model': data[model_key],
                    'model_name': model_name,
                    'category': data['category'],
                    'index': data['index']
                }
                
                # Write the new format
                f_out.write(json.dumps(new_data) + '\n')

def process_directory(input_dir, output_dir):
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)
    
    # Walk through all subdirectories
    for root, dirs, files in os.walk(input_dir):
        # Calculate relative path from input directory
        rel_path = os.path.relpath(root, input_dir)
        
        # Create corresponding output directory
        output_subdir = os.path.join(output_dir, rel_path)
        os.makedirs(output_subdir, exist_ok=True)
        
        # Process all JSONL files
        for file in files:
            if file.endswith('.jsonl'):
                input_file = os.path.join(root, file)
                output_file = os.path.join(output_subdir, file)
                process_jsonl_file(input_file, output_file)
```

```python

# Define input and output directories
input_dir = '/Users/maxschaffelder/Desktop/Thesis/Data/Finetuning/Augmented'
output_dir = '/Users/maxschaffelder/Desktop/Thesis/Data/Finetuning/synthetic'

# Process all files
process_directory(input_dir, output_dir)
```
