# Import and load sBERT embeddings

```python
def load_embeddings(path):
    embeddings_dict = {
        'sentences': [],
        'embeddings': []
    }
    
    with open(path, 'r') as f:
        for line in f:
            data = json.loads(line)
            embeddings_dict['sentences'].append(data['sentence'])
            embeddings_dict['embeddings'].append(np.array(data['embedding']))
            
    # Convert list of embeddings to a numpy array
    embeddings_dict['embeddings'] = np.array(embeddings_dict['embeddings'])
    return embeddings_dict


sbert_embeddings = {
    "Human":{
        "dolly_train_1": load_embeddings("../../Data/analysis/sentence_embeddings/Human/dolly_train_1_human_embeddings.jsonl"),
        "dolly_train_2": load_embeddings("../../Data/analysis/sentence_embeddings/Human/dolly_train_2_human_embeddings.jsonl"),
        "dolly_train_3": load_embeddings("../../Data/analysis/sentence_embeddings/Human/dolly_train_3_human_embeddings.jsonl"),
        "dolly_train_4": load_embeddings("../../Data/analysis/sentence_embeddings/Human/dolly_train_4_human_embeddings.jsonl")
    },
    "Small":{
        "dolly_train_1": load_embeddings("../../Data/analysis/sentence_embeddings/Small/Llama/dolly_train_1_Llama_embeddings.jsonl"),
        "dolly_train_2": load_embeddings("../../Data/analysis/sentence_embeddings/Small/Llama/dolly_train_2_Llama_embeddings.jsonl"),
        "dolly_train_3": load_embeddings("../../Data/analysis/sentence_embeddings/Small/Llama/dolly_train_3_Llama_embeddings.jsonl"),
        "dolly_train_4": load_embeddings("../../Data/analysis/sentence_embeddings/Small/Llama/dolly_train_4_Llama_embeddings.jsonl"),
        "dolly_train_2": load_embeddings("../../Data/analysis/sentence_embeddings/Small/Other/dolly_train_2_Gemma_embeddings.jsonl"),
        "dolly_train_3": load_embeddings("../../Data/analysis/sentence_embeddings/Small/Other/dolly_train_3_Qwen2_embeddings.jsonl"),
        "dolly_train_4": load_embeddings("../../Data/analysis/sentence_embeddings/Small/Other/dolly_train_4_Falcon3-10B-Instruct_embeddings.jsonl")
    },
    "Medium":{
        "dolly_train_1": load_embeddings("../../Data/analysis/sentence_embeddings/Medium/Llama/dolly_train_1_Llama_embeddings.jsonl"),
        "dolly_train_2": load_embeddings("../../Data/analysis/sentence_embeddings/Medium/Llama/dolly_train_2_Llama_embeddings.jsonl"),
        "dolly_train_3": load_embeddings("../../Data/analysis/sentence_embeddings/Medium/Llama/dolly_train_3_Llama_embeddings.jsonl"),
        "dolly_train_4": load_embeddings("../../Data/analysis/sentence_embeddings/Medium/Llama/dolly_train_4_Llama_embeddings.jsonl"),
        "dolly_train_2": load_embeddings("../../Data/analysis/sentence_embeddings/Medium/Other/dolly_train_2_Mistral_embeddings.jsonl"),
        "dolly_train_3": load_embeddings("../../Data/analysis/sentence_embeddings/Medium/Other/dolly_train_3_Qwen_embeddings.jsonl"),
        "dolly_train_4": load_embeddings("../../Data/analysis/sentence_embeddings/Medium/Other/dolly_train_4_Command_embeddings.jsonl")
    },
    "Large": {
        "dolly_train_1": load_embeddings("../../Data/analysis/sentence_embeddings/Large/Llama/dolly_train_1_Llama_embeddings.jsonl"),
        "dolly_train_2": load_embeddings("../../Data/analysis/sentence_embeddings/Large/Llama/dolly_train_2_Llama_embeddings.jsonl"),
        "dolly_train_3": load_embeddings("../../Data/analysis/sentence_embeddings/Large/Llama/dolly_train_3_Llama_embeddings.jsonl"),
        "dolly_train_4": load_embeddings("../../Data/analysis/sentence_embeddings/Large/Llama/dolly_train_4_Llama_embeddings.jsonl"),
        "dolly_train_2": load_embeddings("../../Data/analysis/sentence_embeddings/Large/Other/dolly_train_2_Gemini_embeddings.jsonl"),
        "dolly_train_3": load_embeddings("../../Data/analysis/sentence_embeddings/Large/Other/dolly_train_3_GPT_embeddings.jsonl"),
        "dolly_train_4": load_embeddings("../../Data/analysis/sentence_embeddings/Large/Other/dolly_train_4_Deepseek_embeddings.jsonl")
    }
}
```

# Create t-SNE embeddings

```python
# Initialize t-SNE
tsne = TSNE(n_components=2, perplexity=30, max_iter=1000, random_state=42)

def save_tsne_embeddings(embeddings_dict, output_path):
    # Create directory if it doesn't exist
    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    
    # Apply t-SNE transformation
    tsne_embeddings = tsne.fit_transform(embeddings_dict['embeddings'])
    
    # Save embeddings and sentences to jsonl file
    with open(output_path, 'w') as f:
        for sentence, embedding in zip(embeddings_dict['sentences'], tsne_embeddings):
            data = {
                'sentence': sentence,
                'tsne_embedding': embedding.tolist()
            }
            f.write(json.dumps(data) + '\n')

```

```python
# Process human embeddings
print("Processing human embeddings...")
for dolly_version, embeddings in tqdm(human_embeddings.items()):
    output_path = f"../../Data/analysis/tsne_embeddings/Human/{dolly_version}_human_tsne.jsonl"
    save_tsne_embeddings(embeddings, output_path)
    
print("Done! T-SNE embeddings saved to ../../Data/analysis/tsne_embeddings/")
```
