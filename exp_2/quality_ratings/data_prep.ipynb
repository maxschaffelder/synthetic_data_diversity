```python
import os
import json
```

# Curate dataset for quality judge

```python
model_size = 'small'

refusalbench_responses = {}

# Load data from refusalbench_only folder
refusalbench_only_path = f'/Users/maxschaffelder/Desktop/Thesis/data/exp_2/outputs/{model_size}/rb'
for filename in os.listdir(refusalbench_only_path):
    if filename.endswith('.jsonl'):
        key = filename.replace('.jsonl', '')
        filepath = os.path.join(refusalbench_only_path, filename)
        
        # Load JSONL data
        with open(filepath, 'r') as f:
            data = [json.loads(line) for line in f]
        
        refusalbench_responses[key] = data
        print(f"Loaded {len(data)} responses for {key}")

refusalbench_jailbreak_responses = {}

# Load data from refusalbench_jailbreak folder
refusalbench_jailbreak_path = f'/Users/maxschaffelder/Desktop/Thesis/data/exp_2/outputs/{model_size}/rb_jb'
for filename in os.listdir(refusalbench_jailbreak_path):
    if filename.endswith('.jsonl'):
        key = filename.replace('.jsonl', '')
        filepath = os.path.join(refusalbench_jailbreak_path, filename)
        
        # Load JSONL data
        with open(filepath, 'r') as f:
            data = [json.loads(line) for line in f]
        
        refusalbench_jailbreak_responses[key] = data
        print(f"Loaded {len(data)} responses for {key}")


```

```python
with open ('/Users/maxschaffelder/Desktop/Thesis/data/exp_2/inputs/quality_judge_input_template_v2.txt', 'r') as file:
    judge_instruction = file.read()

#[Question] 
# {question}  

#[The Start of the Assistant's Answer] 
# {answer} 

refusalbench_judge_input = {}

for key, content in refusalbench_responses.items():

    judge_inputs = []

    for line in content:
        line_dict = {}
        judge_input = judge_instruction.replace("{question}", line["instruction"]).replace("{answer}", line["response_model"])
        line_dict["judge_input"] = judge_input
        line_dict["category"] = line["category"]
        line_dict["response_model"] = line["response_model"]
        line_dict["instruction"] = line["instruction"]
        judge_inputs.append(line_dict)
    
    refusalbench_judge_input[key] = judge_inputs


refusalbench_jailbreak_judge_input = {}

for key, content in refusalbench_jailbreak_responses.items():
    judge_inputs = []

    for line in content:
        line_dict = {}
        judge_input = judge_instruction.replace("{question}", line["instruction"]).replace("{answer}", line["response_model"])
        line_dict["judge_input"] = judge_input
        line_dict["category"] = line["category"]
        line_dict["response_model"] = line["response_model"]
        line_dict["instruction"] = line["instruction"]
        judge_inputs.append(line_dict)
    
    refusalbench_jailbreak_judge_input[key] = judge_inputs
```

```python
# Create directories if they don't exist
os.makedirs(f"/Users/maxschaffelder/Desktop/Thesis/data/exp_2/quality_judge/inputs/{model_size}/rb", exist_ok=True)
os.makedirs(f"/Users/maxschaffelder/Desktop/Thesis/data/exp_2/quality_judge/inputs/{model_size}/rb_jb", exist_ok=True)

# Save refusalbench_judge_input contents
for key, content in refusalbench_judge_input.items():
    filepath = os.path.join(f"/Users/maxschaffelder/Desktop/Thesis/data/exp_2/quality_judge/inputs/{model_size}/rb", f"{key}.jsonl")
    with open(filepath, 'w') as f:
        for judge_input in content:
            f.write(json.dumps(judge_input) + "\n")

# Save refusalbench_jailbreak_judge_input contents
for key, content in refusalbench_jailbreak_judge_input.items():
    filepath = os.path.join(f"/Users/maxschaffelder/Desktop/Thesis/data/exp_2/quality_judge/inputs/{model_size}/rb_jb", f"{key}.jsonl")
    with open(filepath, 'w') as f:
        for judge_input in content:
            f.write(json.dumps(judge_input) + "\n")
```
