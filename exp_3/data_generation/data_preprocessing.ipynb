# Prepare data for pairwise ranking

```python
# This is using placeholder paths for now
data_paths = {
    "llama": "/Users/maxschaffelder/Desktop/Thesis/data/exp_3/generated/small/summaries_llama.jsonl",
    "qwen": "/Users/maxschaffelder/Desktop/Thesis/data/exp_3/generated/small/summaries_qwen.jsonl",
    "falcon": "/Users/maxschaffelder/Desktop/Thesis/data/exp_3/generated/small/summaries_falcon.jsonl",
    "gemma": "/Users/maxschaffelder/Desktop/Thesis/data/exp_3/generated/small/summaries_gemma.jsonl"
}

paired_summaries = []
models_A = []
models_B = []

# Open all files and read their contents
data = {}
for model_name, path in data_paths.items():
    with open(path, 'r') as f:
        data[model_name] = [json.loads(line) for line in f]

# Iterate through all entries (assuming all files have same number of entries)
for i in range(len(data[list(data_paths.keys())[0]])):

    # Get summaries for each model at this index
    summaries = {model: data[model][i]['response_model'] for model in data_paths.keys()}


    # Create pairs of summaries
    models = list(data_paths.keys())
    for j in range(len(models)):
        for k in range(j + 1, len(models)):
            model_A = models[j]
            model_B = models[k]

            
            paired_summaries.append(
                "Summary A: " + summaries[model_A] + "\n" +
                "Summary B: " + summaries[model_B]
            )

            models_A.append(model_A)
            models_B.append(model_B)

            model_A, model_B = model_B, model_A

            paired_summaries.append(
                "Summary A: " + summaries[model_A] + "\n" +
                "Summary B: " + summaries[model_B]
            )

            models_A.append(model_A)
            models_B.append(model_B)
```

```python
# This is using placeholder paths for now
data_paths = {
    "llama": "/Users/maxschaffelder/Desktop/Thesis/data/exp_3/generated/small/summaries_llama.jsonl",
    "qwen": "/Users/maxschaffelder/Desktop/Thesis/data/exp_3/generated/small/summaries_qwen.jsonl",
    "falcon": "/Users/maxschaffelder/Desktop/Thesis/data/exp_3/generated/small/summaries_falcon.jsonl",
    "gemma": "/Users/maxschaffelder/Desktop/Thesis/data/exp_3/generated/small/summaries_gemma.jsonl"
}

paired_summaries = []
models_g = []
models_h = []

# Open all files and read their contents
data = {}
for model_name, path in data_paths.items():
    with open(path, 'r') as f:
        data[model_name] = [json.loads(line) for line in f]

# Iterate through all entries (assuming all files have same number of entries)
for i in range(len(data[list(data_paths.keys())[0]])):

    # Get summaries for each model at this index
    summaries = {model: data[model][i]['response_model'] for model in data_paths.keys()}


    # Create pairs of summaries
    models = list(data_paths.keys())
    for j in range(len(models)):
        for k in range(j + 1, len(models)):
            model_g = models[j]
            model_h = models[k]

            
            paired_summaries.append(
                "Summary G: " + summaries[model_g] + "\n" +
                "Summary H: " + summaries[model_h]
            )

            models_g.append(model_g)
            models_h.append(model_h)

            model_g, model_h = model_h, model_g

            paired_summaries.append(
                "Summary G: " + summaries[model_g] + "\n" +
                "Summary H: " + summaries[model_h]
            )

            models_g.append(model_g)
            models_h.append(model_h)
```

```python
# Create DataFrame from the lists
df_pairs = pd.DataFrame({
    'paired_summaries': paired_summaries,
    'model_g': models_g,
    'model_h': models_h
})

# Export to JSONL file
output_path = "/Users/maxschaffelder/Desktop/Thesis/data/exp_3/pairwise_data/pairwise_ranking_input_small.jsonl"
with open(output_path, 'w') as f:
    for _, row in df_pairs.iterrows():
        json_line = {
            'summary': row['paired_summaries'],
            'model_g': row['model_g'],
            'model_h': row['model_h']
        }
        f.write(json.dumps(json_line) + '\n')

print(f"Exported {len(df_pairs)} pairs to {output_path}")
```

# Some dataset analysis

```python
# Read the JSONL file and store in a variable
with open("/Users/maxschaffelder/Desktop/Thesis/data/exp_3/cnn_dailymail_sampled_100.jsonl", 'r') as f:
    generated_summaries = [json.loads(line)['highlights'] for line in f]


# Calculate word and sentence lengths for highlights
word_lengths =  pd.Series([len(summary.split()) for summary in generated_summaries])
sentence_lengths = pd.Series([len(summary.split('.')) for summary in generated_summaries])

# Print statistics for word lengths
print("Length in words:")
print(f"Mean: {word_lengths.mean():.2f}")
print(f"Median: {word_lengths.median():.2f}")
print(f"Max: {word_lengths.max()}")
print(f"Min: {word_lengths.min()}")
print(f"Number of samples above 100 words: {len(word_lengths[word_lengths > 100])} out of {len(word_lengths)}")

print("\nLength in sentences Statistics:")
print(f"Mean: {sentence_lengths.mean():.2f}")
print(f"Median: {sentence_lengths.median():.2f}")
print(f"Max: {sentence_lengths.max()}")
print(f"Min: {sentence_lengths.min()}")

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.hist(word_lengths, bins=15, edgecolor='black')
plt.title('Distribution of Word Lengths in Summaries')
plt.xlabel('Number of Words')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)
plt.show()
```
