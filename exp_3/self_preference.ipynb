# 3.2.3 Experiment 3: Impact of synthetic data diversity on self-preference bias
To understand the impact of synthetic data diversity on self-preference bias, I will first use the target model
as well as three other LLMs to generate summaries based on the CNN/DailyMail dataset [44], which will
result in four summaries. 

Then, I will use the fine-tuned models from experiment 1 (3.2.1) as well as the unaltered model (and the three non-target models?) as control conditions to both pairwise-rank these summaries as well as provide absolute ratings for their performance.


Finally, to understand whether and how self-preference bias is affected by model fine-tuning on synthetic data, I will compare the average ratings given to each summary source-model given by all evaluator-models.

# Pairwise ranking

For direct rankings, I will ask the model which of the two outputs is of higher
quality, and normalize the output probability of the tokens indicative of the two options. To avoid
ordering bias, this will be done twice for each pair, with the options presented in the opposite order.

# Absolute rating
To create absolute ratings that are comparable among different outputs, I will first
let each model rate the given output on a Likert-scale from one to five and calculate the final rating to
be the average of the five output possibilities weighted by their respective probability. To make scores
pairwise comparable, they will be normalized by dividing by the sum of the two to-be-compared scores.
