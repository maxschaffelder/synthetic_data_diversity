```python
import pandas as pd
import json
```

```python
df = pd.read_pickle("/Users/maxschaffelder/Desktop/Thesis/Data/Self_preference/cnn_dailymail_df.pkl")
df.head()
```

```python
# Randomly sample 1000 rows from the dataframe
df_sampled = df.sample(n=100, random_state=42)

df_sampled.head()

# Convert DataFrame to JSONL format and save
output_path = "/Users/maxschaffelder/Desktop/Thesis/Data/Self_preference/cnn_dailymail_sampled_100.jsonl"
df_sampled.to_json(output_path, orient='records', lines=True)
```

# Some dataset analysis

```python
# Calculate word and sentence lengths for highlights
word_lengths = df_sampled['highlights'].str.split().str.len()
sentence_lengths = df_sampled['highlights'].str.split('.').str.len()

# Print statistics for word lengths
print("Length in words:")
print(f"Mean: {word_lengths.mean():.2f}")
print(f"Median: {word_lengths.median():.2f}")
print(f"Max: {word_lengths.max()}")
print(f"Min: {word_lengths.min()}")
print(f"Number of samples above 100 words: {len(word_lengths[word_lengths > 100])} out of {len(word_lengths)}")

print("\nLength in sentences Statistics:")
print(f"Mean: {sentence_lengths.mean():.2f}")
print(f"Median: {sentence_lengths.median():.2f}")
print(f"Max: {sentence_lengths.max()}")
print(f"Min: {sentence_lengths.min()}")
```

```python
# Read the JSONL file and store in a variable
with open("/Users/maxschaffelder/Desktop/Thesis/data/exp_3/generated/small/summaries_llama_vanilla_longer.jsonl", 'r') as f:
    generated_summaries = [json.loads(line)['response_model'] for line in f]

# Calculate word and sentence lengths for generated summaries
gen_word_lengths = pd.Series([len(summary.split()) for summary in generated_summaries])
gen_sentence_lengths = pd.Series([len(summary.split('.')) for summary in generated_summaries])

# Print statistics for word lengths
print("Generated Summaries - Length in words:")
print(f"Mean: {gen_word_lengths.mean():.2f}")
print(f"Median: {gen_word_lengths.median():.2f}")
print(f"Max: {gen_word_lengths.max()}")
print(f"Min: {gen_word_lengths.min()}")
print(f"Number of samples above 100 words: {len(gen_word_lengths[gen_word_lengths > 100])} out of {len(gen_word_lengths)}")

print("\nGenerated Summaries - Length in sentences:")
print(f"Mean: {gen_sentence_lengths.mean():.2f}")
print(f"Median: {gen_sentence_lengths.median():.2f}")
print(f"Max: {gen_sentence_lengths.max()}")
print(f"Min: {gen_sentence_lengths.min()}")
```
